```python
# imports
import torch
from transformers import pipeline
from diffusers import DiffusionPipeline
from datasets import load_dataset
import soundfile as sf
from IPython.display import Audio
```

```python
# sentiment analysis
classifier = pipeline("sentiment-analysis")
result = classifier('I am ok playround AI and Agents')
print(result)
```

```python
# named entity recognition
#ner = pipeline('ner', model='dbmdz/bert-large-cased-finetuned-conll03-english',grouped_entities=True) #, device='cuda'
ner = pipeline('ner', grouped_entities=True) #, device='cuda'
result = ner('Narendar modi is the 14th prime minister of india')
print(result)
```

```python
# Question answering with context
question_answerer = pipeline("question-answering")# , device = 'cuda'
context = "Manmohan Singh was the 13th prime minister of India. He was followed by Narendra Modi, who became the 14th prime minister."
result = question_answerer(question="Who was the 13th prime minister of India?", context=context)
print(result)
```

```python
# Text summarization

summarizer = pipeline("summarization")
result = summarizer(
    """
    Hugging Face is a popular open-source platform that provides powerful tools and models for Natural Language Processing (NLP).
It simplifies working with state-of-the-art Transformer models through easy-to-use APIs and pre-trained resources.

Summary of Hugging Face & the NLP Pipeline:

Pre-trained Models: Offers thousands of ready-to-use Transformer models (BERT, GPT, T5, etc.) for tasks like text classification, translation, and summarization.

Tokenizers: Efficient tokenization tools that convert text into model-understandable numerical representations.

Pipeline API: A high-level interface that allows users to perform NLP tasks in just a few lines of code (e.g., pipeline("sentiment-analysis")).

Datasets Hub: Provides a large collection of pre-processed datasets for training, fine-tuning, and experimentation.

Model Hub: A central place to store, share, and version models, fostering collaboration across the community.

Fine-tuning Support: Enables users to customize pre-trained models on specific datasets with minimal code.
    """
)
print(result)
```

```python
#translation
#translator = pipeline("translation_en_to_fr")
translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-hi")
result = translator("Hugging Face is a great platform to learn NLP!")
print(result)
```

```python
# classfication
classifier = pipeline("zero-shot-classification")
result = classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"],
)
print(result)
```

```python
# Text Generation
generator = pipeline("text-generation")
result = generator("In this course, we will teach you how to")
print(result)
```

```python
# image generation
from google.colab import userdata
from huggingface_hub import login
hf_token = userdata.get("HF_TOKEN")
if hf_token is None:
    raise ValueError("HF_TOKEN is not set in Colab Secrets!")

login(token=hf_token)
image_gen = DiffusionPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    use_safetensors=True
)

# image_gen.to("cuda")

prompt = "a photograph of an astronaut riding a horse"
image = image_gen(prompt).images[0]
image
```

```python
# Audio Generation

synthesiser = pipeline("text-to-speech", "microsoft/speecht5_tts")
embedding_dataset = load_dataset("Matthijs/cmu-arctic-xvectors", split="validation")
speaker_embedding = torch.tensor(embedding_dataset[0]["xvector"]).unsqueeze(0)

speech = synthesiser("Hi to an AI agent", forward_params={"speaker_embeddings": speaker_embedding})
sf.write("speech.wav", speech["audio"], speech["sampling_rate"])
Audio("speech.wav")
```
